```python
import pandas as pd
import numpy as np

# File load karo
df = pd.read_csv('global_ecommerce_sales.csv')
df['Order_Date'] = pd.to_datetime(df['Order_Date'])
df['Day_Name'] = df['Order_Date'].dt.day_name()

print("File loaded! Rows:", len(df))
```

```python
total_rev = df['Total_Amount'].sum()
total_orders = df['Order_ID'].nunique()
total_units = df['Quantity'].sum()
aov_val = total_rev / total_orders

print(f"Total Revenue : ${total_rev:,.2f}")
print(f"Total Orders  : {total_orders:,}")
print(f"Total Volume  : {total_units:,} units")
print(f"AOV           : ${aov_val:,.2f}")
```

```python
# Country aur Category ka matrix ($K me)
pivot_matrix = df.pivot_table(
    index='Country',
    columns='Category',
    values='Total_Amount',
    aggfunc='sum'
) / 1000

pivot_matrix.round(1).head()
```

```python
import matplotlib.pyplot as plt
import seaborn as sns

plt.figure(figsize=(10, 5))
sns.heatmap(pivot_matrix, annot=True, fmt=".1f", cmap="magma", linewidths=0.5)
plt.title("Market Opportunity Matrix: Country × Category ($K)")
plt.show()
```

```python
pareto_df = df.groupby('Country')['Total_Amount'].sum().sort_values(ascending=False).to_frame()
pareto_df['Cumulative_%'] = (pareto_df['Total_Amount'].cumsum() / pareto_df['Total_Amount'].sum()) * 100
pareto_df.round(2)
```

```python
fig, ax1 = plt.subplots(figsize=(9, 4))
ax2 = ax1.twinx()

ax1.bar(pareto_df.index, pareto_df['Total_Amount'] / 1000, color='#38bdf8')
ax2.plot(pareto_df.index, pareto_df['Cumulative_%'], color='#f43f5e', marker='o')
ax2.axhline(80, color='orange', linestyle='--', label='80% Cutoff')

ax1.set_ylabel("Revenue ($K)")
ax2.set_ylabel("Cumulative %")
plt.xticks(rotation=45)
plt.title("Pareto Analysis: Cumulative Revenue Contribution")
plt.show()
```

```python
days_order = ['Monday', 'Tuesday', 'Wednesday', 'Thursday', 'Friday', 'Saturday', 'Sunday']
dow_rev = df.groupby('Day_Name')['Total_Amount'].sum().reindex(days_order)

plt.figure(figsize=(9, 4))
plt.plot(dow_rev.index, dow_rev.values / 1000, marker='s', color='#8b5cf6', linewidth=2)
for x, y in zip(dow_rev.index, dow_rev.values / 1000):
    plt.annotate(f'${y:.0f}K', (x, y), textcoords="offset points", xytext=(0, 6), ha='center')

plt.ylabel("Revenue ($K)")
plt.title("Weekly Order Velocity (Day of Week)")
plt.grid(True, linestyle='--', alpha=0.5)
plt.show()
```

```python
plt.figure(figsize=(8, 4))
sns.histplot(df['Total_Amount'], kde=True, color='#22c55e', bins=25)
plt.axvline(df['Total_Amount'].median(), color='orange', linestyle='-', label=f"Median: ${df['Total_Amount'].median():,.0f}")
plt.axvline(df['Total_Amount'].mean(), color='red', linestyle='--', label=f"Mean: ${df['Total_Amount'].mean():,.0f}")

plt.xlabel("Order Value ($)")
plt.title("Order Value Distribution (Skewness)")
plt.legend()
plt.show()
```

```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import matplotlib.gridspec as gridspec
import matplotlib.patches as mpatches
import seaborn as sns
from google.colab import files

# --- 1. SETUP CANVAS & WHITE THEME ---
fig = plt.figure(figsize=(19, 13), facecolor='#ffffff')
gs = gridspec.GridSpec(3, 2, height_ratios=[0.42, 1.25, 1.25], hspace=0.35, wspace=0.22)

# Main Title & Subtitle with proper top margin
fig.text(0.5, 0.965, 'GLOBAL E-COMMERCE SALES INTELLIGENCE DASHBOARD',
         fontsize=17, fontweight='bold', ha='center', color='#0f172a')
fig.text(0.5, 0.945, 'Executive Portfolio Performance Overview (Jan 2025 – Nov 2025)',
         fontsize=10.5, ha='center', color='#64748b')

# --- 2. EXECUTIVE KPI CARDS ---
ax_kpi = fig.add_subplot(gs[0, :])
ax_kpi.set_facecolor('#ffffff')
ax_kpi.axis('off')

kpi_data = [
    ("TOTAL REVENUE", "$1,512,253", "#0284c7", "10-Month Gross Sales", "#f0f9ff", "#bae6fd"),
    ("TOTAL ORDERS", "1,200", "#16a34a", "6,008 Units Sold", "#f0fdf4", "#bbf7d0"),
    ("AVG ORDER VALUE", "$1,260.21", "#d97706", "Median Value: $950.42", "#fffbeb", "#fde68a"),
    ("TOP REGION", "USA (13.6%)", "#7c3aed", "$206.1K Gross Revenue", "#faf5ff", "#e9d5ff"),
    ("TOP CATEGORY", "Home Decor", "#e11d48", "$232.3K (AOV $1,442)", "#fff1f2", "#fecdd3")
]

for i, (title, val, stroke_col, sub, bg_col, border_col) in enumerate(kpi_data):
    x_pos = 0.01 + i * 0.20
    rect = mpatches.FancyBboxPatch((x_pos, 0.08), 0.18, 0.84, transform=ax_kpi.transAxes,
                                  boxstyle="round,pad=0.02,rounding_size=0.03",
                                  facecolor=bg_col, edgecolor=border_col, lw=1.5)
    ax_kpi.add_patch(rect)
    ax_kpi.text(x_pos + 0.012, 0.68, title, transform=ax_kpi.transAxes, color='#475569', fontsize=9, fontweight='bold')
    ax_kpi.text(x_pos + 0.012, 0.36, val, transform=ax_kpi.transAxes, color=stroke_col, fontsize=15, fontweight='bold')
    ax_kpi.text(x_pos + 0.012, 0.14, sub, transform=ax_kpi.transAxes, color='#64748b', fontsize=8)

# --- 3. MARKET OPPORTUNITY MATRIX (HEATMAP) ---
ax1 = fig.add_subplot(gs[1, 0])
ax1.set_facecolor('#ffffff')
pivot_matrix = df.pivot_table(index='Country', columns='Category', values='Total_Amount', aggfunc='sum') / 1000
sns.heatmap(pivot_matrix, annot=True, fmt=".0f", cmap="YlGnBu", ax=ax1,
            cbar_kws={'label': 'Revenue ($K)'}, linewidths=0.8, linecolor='#e2e8f0', annot_kws={'fontsize': 8.5})
ax1.set_title("Market Opportunity Matrix: Country × Category Revenue ($K)", color='#0f172a', fontweight='bold', fontsize=11, pad=10)
ax1.tick_params(axis='x', rotation=30, labelsize=8.5)
ax1.tick_params(axis='y', labelsize=8.5)
ax1.set_xlabel("Product Category", fontweight='bold', fontsize=9, color='#334155')
ax1.set_ylabel("Country", fontweight='bold', fontsize=9, color='#334155')

# --- 4. PARETO 80/20 ANALYSIS ---
ax2 = fig.add_subplot(gs[1, 1])
ax2.set_facecolor('#ffffff')
pareto_df = df.groupby('Country')['Total_Amount'].sum().sort_values(ascending=False).to_frame()
pareto_df['Cumulative_%'] = (pareto_df['Total_Amount'].cumsum() / pareto_df['Total_Amount'].sum()) * 100

ax2_twin = ax2.twinx()
bars = ax2.bar(pareto_df.index, pareto_df['Total_Amount'] / 1000, color='#0284c7', alpha=0.9, width=0.65)
ax2_twin.plot(pareto_df.index, pareto_df['Cumulative_%'], color='#ef4444', marker='o', linewidth=2.2, markersize=5)
ax2_twin.axhline(80, color='#d97706', linestyle='--', linewidth=1.5, label='80% Cutoff')

ax2.set_title("Pareto Analysis: Cumulative Revenue Contribution", color='#0f172a', fontweight='bold', fontsize=11, pad=10)
ax2.tick_params(axis='x', rotation=35, labelsize=8.5)
ax2.tick_params(axis='y', labelsize=8.5)
ax2_twin.tick_params(labelsize=8.5)
ax2.set_ylabel("Revenue ($K)", fontweight='bold', fontsize=9, color='#334155')
ax2_twin.set_ylabel("Cumulative %", fontweight='bold', fontsize=9, color='#334155')
ax2.grid(axis='y', linestyle='--', alpha=0.4)
ax2_twin.grid(False)

# --- 5. WEEKLY ORDER VELOCITY ---
ax3 = fig.add_subplot(gs[2, 0])
ax3.set_facecolor('#ffffff')
days_order = ['Monday', 'Tuesday', 'Wednesday', 'Thursday', 'Friday', 'Saturday', 'Sunday']
dow_rev = df.groupby('Day_Name')['Total_Amount'].sum().reindex(days_order)
ax3.plot(dow_rev.index, dow_rev.values / 1000, marker='s', color='#7c3aed', linewidth=2.5, markersize=6)
for x, y in zip(dow_rev.index, dow_rev.values / 1000):
    ax3.annotate(f'${y:.0f}K', (x, y), textcoords="offset points", xytext=(0, 6), ha='center', fontsize=8.5, fontweight='bold', color='#4c1d95')

ax3.set_title("Weekly Order Velocity (Day of Week)", color='#0f172a', fontweight='bold', fontsize=11, pad=10)
ax3.tick_params(axis='x', rotation=25, labelsize=8.5)
ax3.tick_params(axis='y', labelsize=8.5)
ax3.set_ylabel("Revenue ($K)", fontweight='bold', fontsize=9, color='#334155')
ax3.grid(True, linestyle='--', alpha=0.4)

# --- 6. TICKET VALUE SKEWNESS & OUTLIERS ---
ax4 = fig.add_subplot(gs[2, 1])
ax4.set_facecolor('#ffffff')
sns.histplot(df['Total_Amount'], kde=True, ax=ax4, color='#10b981', bins=25, alpha=0.55)
median_v = df['Total_Amount'].median()
mean_v = df['Total_Amount'].mean()
ax4.axvline(median_v, color='#d97706', linestyle='-', linewidth=2, label=f"Median: ${median_v:,.0f}")
ax4.axvline(mean_v, color='#dc2626', linestyle='--', linewidth=2, label=f"Mean: ${mean_v:,.0f}")

ax4.set_title("Order Value Distribution (Skewness & AOV)", color='#0f172a', fontweight='bold', fontsize=11, pad=10)
ax4.tick_params(labelsize=8.5)
ax4.set_xlabel("Order Value ($)", fontweight='bold', fontsize=9, color='#334155')
ax4.set_ylabel("Frequency (Orders)", fontweight='bold', fontsize=9, color='#334155')
ax4.legend(facecolor='#ffffff', edgecolor='#cbd5e1', fontsize=8.5)
ax4.grid(True, linestyle='--', alpha=0.4)

# Margin adjustments
plt.subplots_adjust(top=0.91, bottom=0.06, left=0.06, right=0.94)

# Save image file and download
output_file = 'executive_white_dashboard.png'
plt.savefig(output_file, dpi=300, facecolor='#ffffff')
plt.show()

files.download(output_file)
print("Dashboard saved and downloaded cleanly!")
```
